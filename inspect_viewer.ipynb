# Inspection viewer

See a spike in a curve, click it, and look at the maps and raw data of that frame — 1 model, or 2–5 on the same segment. Guide: [docs/inspection_viewer.md](docs/inspection_viewer.md). Mirrors `inspect_viewer.py`; nothing is computed here.

```python
# 1. Setup -- open this notebook from the folder that contains inspect_viewer.py
import os, sys, glob
assert os.path.exists('inspect_viewer.py'), f'run from the repo folder with inspect_viewer.py, not {os.getcwd()}'
try:
    get_ipython().run_line_magic('matplotlib', 'widget')
except Exception as e:
    print('No interactive plots:', e)
    print('-> pip install ipympl ipywidgets, then Kernel > Restart')
from inspect_viewer import Inspector
```

```python
# 2. Find runs (made with --what inspect). Point INSPECT_ROOT elsewhere if they live in another folder.
INSPECT_ROOT = 'experiments/inspect'
runs_found = sorted(os.path.dirname(p) for p in glob.glob(f'{INSPECT_ROOT}/*/*/*/run.json'))
for i, d in enumerate(runs_found):
    print(i, d)
if not runs_found:
    print(f'no runs under {INSPECT_ROOT}/ -- the next cell falls back to a demo on fake data')
```

```python
# 3. Open the viewer: pick 1-5 runs of the same segment by index
DATA_ROOT = None                   # folder holding <dataset>/ ; None = ./data
if runs_found:
    RUNS = [runs_found[i] for i in (0,)]        # e.g. (0, 3) to compare two models
else:                                           # demo: random maps, tiny generated dataset
    sys.path.insert(0, 'tests')
    from _inspect_fake import make_dataset, make_run
    DATA_ROOT = make_dataset('experiments/inspect_demo/data')
    RUNS = [make_run('experiments/inspect_demo', 'cook', seed=1),
            make_run('experiments/inspect_demo', 'thesis_imu', scale=4.0, seed=2)]
viewer = Inspector(RUNS, data_root=DATA_ROOT)
viewer.show()
```

```python
# 4. Export the current view as a GIF: over all frames, and over the iterations of the current frame
print(viewer.export('experiments/inspect/frames.gif'))
if not viewer.iteration.disabled:
    print(viewer.export('experiments/inspect/iterations.gif', over_iterations=True))
```
