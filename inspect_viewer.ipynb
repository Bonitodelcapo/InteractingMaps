# Inspection viewer

See a spike in a curve, jump to that frame, and look at the model's maps and the raw data of that window — for one model, or up to five on the same segment.

Mirrors `inspect_viewer.py` (class `Inspector`). Nothing is computed here: the notebook only reads files.

**Where the runs come from** (on the cluster, then copy the run folders back):

```
python report/run_experiments.py --what inspect --sequences poster --replay-frames 6,14
python replay_frame.py --run experiments/inspect/<dataset>/<model>/<run> --frame 6
```

The first saves the model state at every frame into `experiments/inspect/`; the second (or `--replay-frames`) replays a frame iteration by iteration. The raw panels and the events/|ω| curves need the dataset under `data/<dataset>/` (or pass `data_root=`).

````python
%matplotlib widget
import glob, os
from inspect_viewer import Inspector

runs_found = sorted(os.path.dirname(p) for p in glob.glob('experiments/inspect/*/*/*/run.json'))
for i, d in enumerate(runs_found):
    print(i, d)
````

## Pick the runs

1 run, or 2–5 runs of the **same** segment (same dataset, segment, t_start, dt, n_frames) — anything else is refused.

- **frame** slider, or click on any curve to jump to the nearest frame.
- **iteration** slider: on when a compared run has a replay of this frame; 0 = the frame's saved state. Runs with fewer iterations hold their last; runs without a replay are titled *no replay*.
- **curves**: any column of `tracking.csv`, `n_events` (events per frame) and `omega_fullrate` (gyro and reference |ω| at full rate).
- Map colours use one fixed scale per quantity for the whole sequence, per model; tick *share scale* to compare magnitudes across models.

````python
RUNS = runs_found[:2]          # pick by index, e.g. [runs_found[i] for i in (0, 3)]
if RUNS:
    viewer = Inspector(RUNS)
    viewer.show()
else:
    print('no inspect runs under experiments/inspect/ -- see the first cell')
````

## Export

The same figure as a GIF (or MP4 by extension): over all frames, or over the iterations of the current frame.

````python
if RUNS:
    print(viewer.export('experiments/inspect/frames.gif'))
    if not viewer.iteration.disabled:
        print(viewer.export('experiments/inspect/iterations.gif', over_iterations=True))
````
